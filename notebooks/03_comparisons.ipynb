# Posterior collapse

During the early training phases of the multimodal variational autoencoder, we realized that setting beta to a fixed value lead to the latent space collapsing. By posterior collapse, we refer to the phenomenon of the learnt distribution q(z|x) being identical to the prior p(z) \sim normal(z; 0,1). This happens when the easiest way to minimize the loss function is to minimize the Kullback-Leibler divergence between q(z|x) and p(z). Hence, as prescribed in the literature (add ref), we introduce the regularization term progressively,. This strategy is called beta annealing. It simply consists of a linear increase of beta from a set value beta_min to another set value beta_max over a set number of epochs.

# Mixing the learnt latent representations

Since we are learning a common latent representation for two modalities, it is important to question what strategy we use to mix the representations learnt by each encoder. We could simply test weights lambda_1 and lambda_2 and set 
mu = lambda_1 mu_rna + lambda_2 mu_atac and sigma = lambda_1 sigma_rna + lambda_2 sigma_atac based on which pair of weights give the best results. A different, and better, strategy is to use an extra linear layer and let the model learn the weights itself. This is what we elect to do.
